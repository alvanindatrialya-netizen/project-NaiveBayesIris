# Library

```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn import metrics
from sklearn.metrics import classification_report
from sklearn.metrics import ConfusionMatrixDisplay
from sklearn.model_selection import train_test_split
```

# Read Data

```python
df_nbiris = pd.read_csv('Dataset_Iris.csv')
df_nbiris
```

```python
df_nbiris.info()
```

```python
df_nbiris['Species'].value_counts()
```

# Independent Dependent Variabel

```python
x = df_nbiris.iloc[:,[1, 2, 3, 4]].values
y = df_nbiris.iloc[:, [5]].values
```

```python
x
```

```python
y
```

# Split Train Test

```python
# Skenario 1: Train 80% : Test 20%
x_train_1, x_test_1, y_train_1, y_test_1 = train_test_split(x, y, test_size=0.2, random_state=0)

# Skenario 2: Train 70% : Test 30%
x_train_2, x_test_2, y_train_2, y_test_2 = train_test_split(x, y, test_size=0.3, random_state=0)

# Skenario 3: Train 60% : Test 40%
x_train_3, x_test_3, y_train_3, y_test_3 = train_test_split(x, y, test_size=0.4, random_state=0)
```

# Create Model & Training

```python
nb_clf1 = GaussianNB()
nb_clf1.fit(x_train_1, y_train_1)

nb_clf2 = GaussianNB()
nb_clf2.fit(x_train_2, y_train_2)

nb_clf3 = GaussianNB()
nb_clf3.fit(x_train_3, y_train_3)
```

# Testing Model

```python
y_prediksi1 = nb_clf1.predict(x_test_1)
y_prediksi2 = nb_clf2.predict(x_test_2)
y_prediksi3 = nb_clf3.predict(x_test_3)
```

```python
print("Prediksi Skenario 1:", y_prediksi1)
print("Prediksi Skenario 2:", y_prediksi2)
print("Prediksi Skenario 3:", y_prediksi3)
```

# Evaluasi Model

```python
print("Test Accuration Skenario 1:", metrics.accuracy_score(y_test_1, y_prediksi1))
print(metrics.classification_report(y_test_1, y_prediksi1))
ConfusionMatrixDisplay.from_predictions(y_test_1, y_prediksi1)
```

```python
print("Test Accuration Skenario 2:", metrics.accuracy_score(y_test_2, y_prediksi2))
print(metrics.classification_report(y_test_2, y_prediksi2))
ConfusionMatrixDisplay.from_predictions(y_test_2, y_prediksi2)
```

```python
print("Test Accuration Skenario 3:", metrics.accuracy_score(y_test_3, y_prediksi3))
print(metrics.classification_report(y_test_3, y_prediksi3))
ConfusionMatrixDisplay.from_predictions(y_test_3, y_prediksi3)
```

# Test New Data

```python
data_baru = [[5.1, 3.5, 1.4, 0.2]]
data_baru
```

```python
predict_data1 = nb_clf1.predict(data_baru)
print("Prediksi Skenario 1 (80:20):", predict_data1)

predict_data2 = nb_clf2.predict(data_baru)
print("Prediksi Skenario 2 (70:30):", predict_data2)

predict_data3 = nb_clf3.predict(data_baru)
print("Prediksi Skenario 3 (60:40):", predict_data3)

```

# Baru
